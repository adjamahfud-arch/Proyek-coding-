```python
#code proyek daftar nama berulang menggunakan while,list,for
nama = []
#sistem penambahan
def daftar_nama(orang):
    for orang in nama:
        print(orang)
#code yang di tampilkan pada pengguna
pilih = "ya"
#code pengulangan
while pilih == "ya":
      orang =input("masukan nama:")
      nama.append(orang)
      pilih =input("mau tambah nama lagi? ya/tidak")
      print("==daftar nama==")
      daftar_nama(orang)

#hasil
print("terimakasih sudah mencoba program saya")
```

```python
#pryek kalkulator sederhana
def kalkulator(angka, angka2, oprasi):
      if oprasi == "+":
         return angka + angka2
      elif oprasi == "-":
         return angka - angka2
      elif oprasi == "/":
         return float(angka / angka2)
      elif oprasi == "*":
         return angka * angka2
#yang diatas adalah sistem hitung
print("==kalkulator sederhana==")

pilih = "ya"
#perulangan
while pilih == "ya":
      angka =float(input("masukan angka pertama:"))
      oprasi =input("pilih(+,-,/,*)")
      angka2 =float(input("masukan angka kedua:"))
      hasil =kalkulator(angka, angka2, oprasi)
      print("-" * 30)
      print(f"hasil: {hasil}")
      print("-" * 30)
      pilih =input("mau hitung lagi? YA/TIDAK")
#hasil
      print("terimakasih sudah memainkan kalkulator sederhana saya")
```
