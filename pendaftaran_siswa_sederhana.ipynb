```python
data_siswa = []

#funcion untuk memberikan data pada list
def siswa(nama, nilai):
    data = {
         "nama":nama,
         "nilai": nilai
}
    data_siswa.append(data)
    print("data berhasil disimpan")
#fungcion untuk menjalankan perintah nomor 2
def nomor_2():
    for orang in data_siswa:
        print(f"nama:{orang['nama']}, nilai:{orang['nilai']}")
#fungtion untuk menjalankan perintah nomor 3
def nomor_3(nilai_baru, nama_baru):
    for tabel in data_siswa:
        if tabel["nama"] == nama_baru:
           tabel["nilai"] = nilai_baru;
           print("data berhasil di ubah")
           return

    print("data tidak ditemukan")
#fungsi untuk menghapus data pada list
def nomor_4(hapus_nama):
   for hapus in data_siswa:
       if hapus["nama"] == hapus_nama:
          data_siswa.remove(hapus)
          print("data berhasil di hapus")
          return
   print("data tidak ditemukan")
#fungsi untuk mencari data nama
def cari_siswa(cari):
    for search in data_siswa:
        if search["nama"] == cari:
           print("data di temukan!")
           print(f"nama:{search['nama']}")
           print(f"nilai:{search['nilai']}")
           return
    print("data tidak ditemukan")









print("====menu data siswa====")
print("1.tambah nama")
print("2.tampilkan nama")
print("3.tambahkan nilai")
print("4.hapus nama")
print("5.keluar")
print("6.cari siswa")

pilih = 0
while pilih != 5:
  try:
    pilih =int(input("pilih nomor yang tertera:"))

  except ValueError:
      print("input harus berupa angka")

      #fungsi untuk menjalankan input nomor pilih

  if pilih == 1:
         try:
           nama =input("masukan nama:").lower()

           nilai =float(input("masukan nilai:"))
           if nilai < 0 or nilai > 100:
              print("nilai tidak boleh lebih dari 100")
           else:
              siswa(nama, nilai)
         except ValueError:
              print("jawaban harus berupa angka ")


  elif pilih == 2:
           print ("===daftar nama==")
           nomor_2()
  elif pilih == 3:
         try:
            nama_baru =input("masukan nama yang mau nilai diganti:").lower()
            nilai_baru =float(input("Masukan niai yangg baru:"))
            if nilai < 0 or nilai > 100:
               print("nilai tidak boleh melebihi 100")
            else:
               nomor_3(nilai_baru, nama_baru)
         except ValueError:
               print("nama tidak boleh angka dan nilai tidak boleh huruf")

  elif pilih == 4:
         try:
            hapus_nama =input("masukan nama yang mau di hapus:").lower()

         except ValueError:
            print("nama harus berupa huruf bukan angka")
         else:
            nomor_4(hapus_nama)
  elif pilih == 5:
           print("keluar dari sistem")

  elif pilih == 6:
         try:
            cari =input("masukan nama yang mau di cari:").lower()
         except ValueError:
            print("nama harus berupa huruf bukan angka")
         else:
            cari_siswa(cari)


print("terimakasih")
```
